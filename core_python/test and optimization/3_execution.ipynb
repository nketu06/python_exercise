```python
# Python Bytecode & Execution
# Python code → Bytecode → Python Virtual Machine (PVM)


# Inspect Bytecode using dis
import dis

def add(a, b):
    return a + b

dis.dis(add)
```

### Bytecode Breakdown

| Bytecode | Meaning |
|----------|---------|
| `LOAD_FAST 0 (a)` | Push variable `a` onto stack |
| `LOAD_FAST 1 (b)` | Push variable `b` onto stack |
| `BINARY_ADD` | Pop 2 values, add them, push result |
| `RETURN_VALUE` | Pop result, return it |

### Why Useful

- See **how Python optimizes** your code
- Debug performance issues
- Understand function behavior at VM level
