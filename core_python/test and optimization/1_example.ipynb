```python
# Why use mocking?

    # 👉 To:
    # avoid slow operations
    # avoid external dependencies
    # test logic in isolation
    # ✅ Basic tool

# Python uses:

from unittest.mock import patch
import requests

# 🔹 Example (without mocking)
def get_data():
    return requests.get("https://api.com").json()

def process():
    data = get_data()
    return data["value"]

# 👉 Problem:
    # Makes real API call ❌
    # ✅ With mocking

from unittest.mock import patch

def test_process():
    with patch("module_name.get_data") as mock_get_data:
        mock_get_data.return_value = {"value": 10}
        
        assert process() == 10

# 👉 No API call ✔
```

```python
# ...existing code...
from unittest.mock import patch
import requests

# Real public testing API
API_URL = "https://jsonplaceholder.typicode.com/todos/1"

def get_data():
    resp = requests.get(API_URL, timeout=5)
    resp.raise_for_status()
    print(resp.json())
    return resp.json()

def process():
    data = get_data()
    return data["title"]

def test_process_mocked():
    # In notebooks, patch __main__.function_name
    with patch("__main__.get_data") as mock_get_data:
        mock_get_data.return_value = {"title": "mocked title"}
        assert process() == "mocked title"

def test_get_data_real_api():
    data = get_data()
    assert "title" in data
    assert isinstance(data["title"], str)

# test_get_data_real_api()
test_process_mocked()
```

```python
# Key idea
#     👉 Replace:
#     real_function → fake_function

# 🔥 Important rule (VERY COMMON MISTAKE)
#     👉 Patch where it is used, not where it is defined
#     patch("module_name.get_data")  ✅

#     NOT:
#     patch("original_file.get_data") ❌

# Mock return value
#     mock_func.return_value = 100

#🔹 Mock side effects
#     mock_func.side_effect = Exception("error")
# 👉 simulate failure

# 🔹 Mock multiple calls
#     mock_func.side_effect = [1, 2, 3]
```

```python
# Using decorator (cleaner)
@patch("module_name.get_data")
def test_process(mock_get_data):
    mock_get_data.return_value = {"value": 10}
    assert process() == 10
```

```python
# Mock object attributes
mock_response = Mock()
mock_response.json.return_value = {"value": 10}
```

```python
# pytest-mock (better way)

# Install:
# pip install pytest-mock

# Use:

def test_process(mocker):
    mocker.patch("module_name.get_data", return_value={"value": 10})
    assert process() == 10

# 👉 Cleaner than patch
```

| Concept      | Meaning           |
| ------------ | ----------------- |
| Mock         | fake object       |
| patch        | replace function  |
| return_value | fake output       |
| side_effect  | simulate behavior |


What is monkeypatch?

- 👉 monkeypatch is a pytest fixture used to:

temporarily change functions, variables, environment
during a test

- 👉 After test → everything is automatically restored

🔥 When to use
 - override functions
 - fake environment variables
 - change configs
 - avoid external calls

```python
def get_value():
    return 10

def test_get_value(monkeypatch):
    monkeypatch.setattr("module.get_value", lambda: 20)

    assert get_value() == 20
```
