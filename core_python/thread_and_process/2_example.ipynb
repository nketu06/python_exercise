### Async vs Multithreading (Quick View)

| Aspect | Async (`asyncio`) | Multithreading (`threading`) |
|---|---|---|
| Model | Single thread, cooperative tasks (`await`) | Multiple OS threads, preemptive scheduling |
| Best for | High-volume I/O (APIs, sockets, many waits) | Blocking I/O libraries that are not async-aware |
| CPU-bound speedup | ❌ | ❌ (GIL in CPython) |
| Context switch cost | Very low | Higher |
| Shared state issues | Fewer race conditions | More locking/synchronization needed |
| Ecosystem fit | Needs async-compatible libraries | Works with many blocking libraries |

**Thumb rule:**
- Use **asyncio** for many concurrent I/O tasks.
- Use **threads** for blocking I/O code that cannot be made async.
- Use **processes** for CPU-bound parallel work.
