Why NOT use print()?  
print("Error happened")

Problems:
- ❌ no log levels
- ❌ no timestamps
- ❌ no persistence
- ❌ hard to filter

| Level    | Purpose       |
| -------- | ------------- |
| DEBUG    | detailed info |
| INFO     | normal events |
| WARNING  | something odd |
| ERROR    | failure       |
| CRITICAL | system crash  |


```python
import logging
logging.debug("debugging")
logging.info("info")
logging.warning("warning")
logging.error("error")
```

```python
logging.basicConfig(
    level=logging.DEBUG,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

# Got it. In Jupyter, basicConfig() often does nothing after logging is already initialized.

logging.debug("debugging")
logging.info("info")
logging.warning("warning")
logging.error("error")
```

```python
import logging

logger = logging.getLogger("my_app") #custome logger
logger.setLevel(logging.INFO)

handler = logging.FileHandler("app.log") #Handlers decide where logs go

formatter = logging.Formatter(
    "%(asctime)s - %(levelname)s - %(message)s"
)

handler.setFormatter(formatter)
logger.addHandler(handler)

logger.info("App started")
logger.info("User login", extra={"user_id": 123})
```

```python
# ...existing code...
import logging

class EnsureUserId(logging.Filter):
    def filter(self, record):
        if not hasattr(record, "user_id"):
            record.user_id = "-"
        return True

logger = logging.getLogger("my_app")
logger.setLevel(logging.INFO)
logger.handlers.clear()      # avoid duplicate logs in notebook reruns
logger.propagate = False

fmt = logging.Formatter(
    "%(asctime)s - %(levelname)s - user_id=%(user_id)s - %(message)s"
)

file_handler = logging.FileHandler("app.log")
file_handler.setFormatter(fmt)
file_handler.addFilter(EnsureUserId())

# console_handler = logging.StreamHandler()  # show in notebook output too
# console_handler.setFormatter(fmt)
# console_handler.addFilter(EnsureUserId())

logger.addHandler(file_handler)
# logger.addHandler(console_handler)

logger.info("App started")
logger.info("User login", extra={"user_id": 123})
# ...existing code...
```
