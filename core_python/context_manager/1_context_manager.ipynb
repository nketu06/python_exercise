````python
# Exception Handling
class MyContext:

    def __enter__(self):
        return self

    def __exit__(self, exc_type, exc, tb):
        print("Handled exception")
        return True

# If exception occurs:
with MyContext():
    1 / 0

# Output:
# Handled exception
# No crash because:
# __exit__ returned True → exception suppressed
````

### contextlib — Cleaner Way (⭐⭐⭐⭐⭐)

#### Overview
- `contextlib` provides utilities to work with context managers in a **clean and pythonic way**
- Instead of writing a full class with `__enter__` and `__exit__`, you can use a **decorator-based approach**
- Makes code shorter, more readable, and easier to maintain

---

#### Traditional Way (Class-based)

```python
class MyContext:
    def __enter__(self):
        print("Enter")
        return self

    def __exit__(self, exc_type, exc, tb):
        print("Exit")

with MyContext():
    print("Inside")

````python
# Cleaner Way (Using contextlib)
from contextlib import contextmanager

@contextmanager
def my_context():
    print("Enter")
    yield
    print("Exit")

with my_context():
    print("Inside")

# ✅ Short
# ✅ Readable
# ✅ Pythonic

# How it Works
# Code before yield → acts like __enter__
# Code after yield → acts like __exit__
````

````python
# Handling Exceptions
from contextlib import contextmanager

@contextmanager
def my_context():
    print("Enter")
    try:
        yield
    except Exception as e:
        print("Exception:", e)
    finally:
        print("Exit")
````

````python
@contextmanager
def open_file(name):
    f = open(name, "r")
    try:
        yield f
    finally:
        f.close()

with open_file("test.txt") as f:
    print(f.read())
````

````python
#   Key Benefits
# No need to define a class
# Less boilerplate
# Easier exception handling
# Better readability
# Ideal for quick utilities


#   When to Use
# Simple resource management (files, locks, DB connections)
# Temporary setup/cleanup logic
# When class-based context manager feels overkill
````

| Approach                     | Code Size | Readability | Use Case      |
| ---------------------------- | --------- | ----------- | ------------- |
| Class (`__enter__/__exit__`) | High      | Medium      | Complex logic |
| `contextlib` decorator       | Low       | High        | Most cases    |


🧠 9️⃣ When to Use  

Use context managers when:  
managing resources  
opening files  
acquiring locks  
handling connections  
temporary state changes

with block = safe zone  
enter → run → cleanup
