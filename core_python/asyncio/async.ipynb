What is a Task?  

A Task wraps a coroutine and schedules it on the event loop.

````python
import asyncio

async def work():
    print("Working")
    await asyncio.sleep(1)
    print("Done")

async def main():
    task = asyncio.create_task(work())
    await task

asyncio.run(main())
````

gather() vs create_task()
gather()

Runs multiple coroutines

Waits for all

Returns results

create_task()

Starts task immediately

You manage it manually

````python
async def work(n):
    print(f"Start {n}")
    await asyncio.sleep(2)
    print(f"End {n}")

async def main():
    await asyncio.gather(work(1), work(2))

asyncio.run(main())
````

````python
# Task Cancellation
async def work():
    try:
        await asyncio.sleep(5)
    except asyncio.CancelledError:
        print("Task cancelled")

async def main():
    task = asyncio.create_task(work())
    await asyncio.sleep(1)
    task.cancel()

asyncio.run(main())
````

````python
# timeout
async def work():
    await asyncio.sleep(5)

async def main():
    try:
        await asyncio.wait_for(work(), timeout=2)
    except asyncio.TimeoutError:
        print("Timeout!")

asyncio.run(main())
````

````python
# Background Tasks (IMPORTANT)
async def main():
    asyncio.create_task(work())  # runs in background
    print("Main done")

# ⚠️ If main exits early → task may be killed
````

````python
import asyncio

async def task(n):
    await asyncio.sleep(n)
    return n

async def main():
    tasks = [asyncio.create_task(task(i)) for i in [1, 2, 3]]

    done, pending = await asyncio.wait(
        tasks,
        return_when=asyncio.FIRST_COMPLETED
    )

    print("Done:", [t.result() for t in done])
    print("Pending:", len(pending))

asyncio.run(main())
````

| Feature            | asyncio.wait()       | asyncio.gather()   |
| ------------------ | -------------------- | ------------------ |
| Return type        | (done, pending) sets | results list       |
| Order preserved    | No                   | Yes                |
| Partial completion | Yes                  | No                 |
| Timeout support    | Yes                  | No                 |
| Exception handling | Flexible             | Raises immediately |
| Ease of use        | Lower                | Higher             |



- `asyncio.wait()` is a low-level API for running multiple async tasks concurrently.
- It provides more control compared to `asyncio.gather()`.
- It works with **Task/Future objects**, not raw coroutines.
