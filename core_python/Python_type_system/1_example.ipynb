 - ✔ better tooling (IDE, autocomplete)
 - ✔ static analysis (mypy)
 - ✔ fewer bugs

```python
# Optional and Union

from typing import Optional

def get_name() -> Optional[str]:
    return None

# Equivalent: str | None


```

```python
# What Union means
# Union[A, B] = value can be either type A OR type B

from typing import Union

def add(x: Union[int, float], y: Union[int, float]) -> float:
    return x + y
```

```python
# You don’t even need Union anymore:

# def add(x: int | float, y: int | float) -> float:
#     return x + y
```

```python
# Typed Dict?
from typing import TypedDict

class User(TypedDict):
    name: str
    age: int

# Now:user: User = {"name": "Nishant", "age": 25}

# What it does
# Ensures dictionary has:
#     specific keys
#     specific value types
# Used by tools like mypy, IDEs
# ⚠️ At runtime → it's still just a normal dict
```

```python
from typing import TypedDict
class User(TypedDict, total=False):
    name: str
    age: int

# 👉 Now both keys are optional
```

```python
from typing import TypedDict, NotRequired

class User(TypedDict):
    name: str
    age: NotRequired[int]
# name required, age optional
```

#### When to use

Use TypedDict when:

- You’re working with JSON-like data
- API responses
- configs
- dictionaries with fixed schema

Avoid if:

- You need behavior → use class/dataclass instead
- Structure is dynamic → use normal dict

| Feature        | TypedDict | dataclass          |
| -------------- | --------- | ------------------ |
| Type safety    | ✅         | ✅                  |
| Runtime object | ❌ (dict)  | ✅                  |
| Methods        | ❌         | ✅                  |
| Best for       | JSON/API  | structured objects |


Protocol  
It checks “does this object have what I need?”  
NOT “is this object from a specific class?”

```python
# Without Protocol (bad design)
class MySQLDB:
    def fetch(self):
        return "data from mysql"

class APIClient:
    def get_data(self):
        return "data from api"
```

```python
# With Protocol (clean design)
from typing import Protocol

class DataSource(Protocol):
    def fetch(self) -> str:
        ...

class MySQLDB:
    def fetch(self):
        return "data from mysql"

class APIClient:
    def fetch(self):
        return "data from api"
```

```python
# Use it anywhere
def process_data(source: DataSource):
    data = source.fetch()
    print(data)
```

```python
# Final (constant variable)
# 👉 Means: this value should NOT be changed

from typing import Final

PI: Final = 3.14

# Literal (fixed allowed values)
# 👉 Means: variable can only be specific exact values

from typing import Literal

def set_mode(mode: Literal["read", "write"]):
    print(mode)
```

enerics = write code once, work with any type  

Instead of fixing a type (int, str), you use a placeholder type.

```python
# Without Generics
def get_first(items: list[int]) -> int:
    return items[0]

# with Generics
from typing import TypeVar
T = TypeVar("T")
def get_first(items: list[T]) -> T:
    return items[0]

# get_first([1, 2, 3])        # int
# get_first(["a", "b"])       # str
# get_first([True, False])    # bool

```

```python
from typing import TypeVar

K = TypeVar("K")
V = TypeVar("V")

def get_value(data: dict[K, V], key: K) -> V:
    return data[key]
```

```python
from typing import Protocol, TypeVar

T = TypeVar("T")

class Loader(Protocol[T]):
    def load(self) -> T:
        ...
```
