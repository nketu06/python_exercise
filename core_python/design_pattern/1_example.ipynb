```python
# singleton

class singleton:
    isinstance = None

    def __new__(cls):
        if cls.isinstance is None:
            cls.isinstance = super().__new__(cls)
        return cls.isinstance
    

s1 = singleton()
s2 = singleton()
print(id(s1),id(s2))      
```
