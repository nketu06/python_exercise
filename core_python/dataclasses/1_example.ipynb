Dataclasses Deep Dive   

(Clean, powerful, production-level data modeling)

```python
# What Problem Dataclasses Solve
# Without dataclass:

class User:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __repr__(self):
        return f"User({self.name}, {self.age})"

# Boilerplate ❌

# ✅ With Dataclass
from dataclasses import dataclass

@dataclass
class User:
    name: str
    age: int

# ✔ auto __init__
# ✔ auto __repr__
# ✔ auto __eq__
```

```python
# Default Values
@dataclass
class User:
    name: str
    age: int = 18

# ⚠️ Rule:
# Non-default fields must come before default fields
```

```python
# Mutable Defaults Trap
# ❌ Wrong:

@dataclass
class A:
    items: list = []

# Shared list across instances 😈
# ✅ Correct:

from dataclasses import field

@dataclass
class A:
    items: list = field(default_factory=list)
```

```python
# field() Advanced Usage
# field(default=..., init=..., repr=..., compare=...)

# Example:
from dataclasses import dataclass,field

@dataclass
class User:
    name: str
    password: str = field(repr=False)

# password hidden in print

print(User(name='ketu',password='k'))
```

```python
# Immutability (frozen=True)
@dataclass(frozen=True)
class User:
    name: str
u = User("Nishant")
u.name = "New"  # Error

# ✔ Useful for:
# configs
# constants
# thread safety
```

```python
# Post Initialization
@dataclass
class User:
    name: str
    age: int

    def __post_init__(self):
        print("Initialized")

# Runs AFTER __init__.
# Use case:
# validation
# derived fields
```

```python
@dataclass
class User:
    first: str
    last: str

    def __post_init__(self):
        self.full = f"{self.first} {self.last}"
```

```python
# Dataclass + Slots 
@dataclass(slots=True)
class User:
    name: str
    age: int

# Benefits:
# ✔ less memory
# ✔ faster access
# ✔ no __dict__
```

```python
# Comparison & Ordering
@dataclass(order=True)
class User:
    age: int
User(20) < User(30)  #✔ True
```

```python
# Convert to Dict
from dataclasses import asdict

asdict(user)

# Useful for:
# APIs
# JSON serialization
```
